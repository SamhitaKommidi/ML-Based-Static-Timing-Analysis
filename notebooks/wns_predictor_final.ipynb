```python

# ================================================================
# PHYSICS-INFORMED RANDOM FOREST
# Complete training, validation, test evaluation and diagnostics
# ================================================================

import os
import json
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    confusion_matrix,
    classification_report,
    precision_score,
    recall_score,
    f1_score
)
from sklearn.model_selection import GroupShuffleSplit

warnings.filterwarnings("ignore", category=FutureWarning)

SEED = 42
TARGET = "worst_slack_ns"
CLOCK = "clock_period_ns"
SAFETY_MARGIN = 0.175
OUTPUT_DIR = "/content/rf_final_results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ================================================================
# 1. LOAD DATASETS
# ================================================================

train = pd.read_csv("/content/model_a_train_combined.csv")
val = pd.read_csv("/content/model_a_validation_combined.csv")
test = pd.read_csv("/content/model_a_test_combined.csv")

print("=" * 70)
print("DATASET SUMMARY")
print("=" * 70)

for name, df in [
    ("Training", train),
    ("Validation", val),
    ("Test", test)
]:
    print(
        f"{name:12s}: {len(df):4d} observations | "
        f"{df['architecture'].nunique():3d} architectures | "
        f"{df['rtl_sha256'].nunique():3d} RTL hashes"
    )

assert list(train.columns) == list(val.columns) == list(test.columns)

for key in ["architecture", "rtl_sha256"]:
    assert set(train[key]).isdisjoint(val[key])
    assert set(train[key]).isdisjoint(test[key])
    assert set(val[key]).isdisjoint(test[key])

print("\nArchitecture and RTL-hash split isolation: PASS")

# ================================================================
# 2. FEATURE EXTRACTION
# ================================================================

# All columns before the first STA target are input/metadata.
target_start = train.columns.get_loc(
    "critical_path_data_arrival_ns"
)

input_columns = list(train.columns[:target_start])

feature_columns = [
    col for col in input_columns
    if pd.api.types.is_numeric_dtype(train[col])
    and col != "sample_id"
]

assert CLOCK in feature_columns

# Physics-informed model excludes clock period from ML features.
# Clock period is used analytically during WNS reconstruction.

structural_features = [
    col for col in feature_columns
    if col != CLOCK
]

X_train = train[structural_features].copy()
X_val = val[structural_features].copy()
X_test = test[structural_features].copy()

T_train = train[CLOCK].to_numpy()
T_val = val[CLOCK].to_numpy()
T_test = test[CLOCK].to_numpy()

y_train = train[TARGET].to_numpy()
y_val = val[TARGET].to_numpy()
y_test = test[TARGET].to_numpy()

# Effective timing requirement
D_train = T_train - y_train
D_val = T_val - y_val
D_test = T_test - y_test

assert np.isfinite(D_train).all()
assert np.isfinite(D_val).all()
assert np.isfinite(D_test).all()

print("\nNumber of structural features:", len(structural_features))
print("Features:")
print(structural_features)

# ================================================================
# 3. MODEL DEFINITION
# ================================================================

rf_model = Pipeline([
    ("imputer", SimpleImputer(
        strategy="median",
        add_indicator=True,
        keep_empty_features=True
    )),
    ("model", RandomForestRegressor(
        n_estimators=300,
        max_depth=12,
        min_samples_leaf=3,
        random_state=SEED,
        n_jobs=-1
    ))
])

# ================================================================
# 4. TRAIN MODEL
# ================================================================

print("\n" + "=" * 70)
print("TRAINING RANDOM FOREST")
print("=" * 70)

rf_model.fit(X_train, D_train)

def predict_wns(model, X, clock_period):
    predicted_delay = model.predict(X)
    return clock_period - predicted_delay

pred_train = predict_wns(rf_model, X_train, T_train)
pred_val = predict_wns(rf_model, X_val, T_val)
pred_test = predict_wns(rf_model, X_test, T_test)

print("Training completed.")

# ================================================================
# 5. REGRESSION METRICS
# ================================================================

def regression_metrics(actual, predicted):
    error = predicted - actual
    abs_error = np.abs(error)

    return {
        "MAE": mean_absolute_error(actual, predicted),
        "RMSE": np.sqrt(mean_squared_error(actual, predicted)),
        "R2": r2_score(actual, predicted),
        "Mean Bias": np.mean(error),
        "Median AE": np.median(abs_error),
        "95th Percentile AE": np.percentile(abs_error, 95),
        "Max AE": np.max(abs_error),
        "Within ±0.1 ns (%)": 100 * np.mean(abs_error <= 0.1),
        "Within ±0.2 ns (%)": 100 * np.mean(abs_error <= 0.2),
        "Within ±0.3 ns (%)": 100 * np.mean(abs_error <= 0.3),
        "Within ±0.5 ns (%)": 100 * np.mean(abs_error <= 0.5),
        "Within ±1.0 ns (%)": 100 * np.mean(abs_error <= 1.0)
    }

regression_results = pd.DataFrame({
    "Training": regression_metrics(y_train, pred_train),
    "Validation": regression_metrics(y_val, pred_val),
    "Test": regression_metrics(y_test, pred_test)
}).T

print("\n" + "=" * 70)
print("REGRESSION PERFORMANCE")
print("=" * 70)

display(regression_results.round(4))

# ================================================================
# 6. VIOLATION CLASSIFICATION
# ================================================================

def classification_metrics(actual, predicted, threshold):
    actual_violation = actual < 0
    predicted_violation = predicted < threshold

    tn, fp, fn, tp = confusion_matrix(
        actual_violation,
        predicted_violation,
        labels=[False, True]
    ).ravel()

    return {
        "Threshold": threshold,
        "Precision": precision_score(
            actual_violation, predicted_violation,
            zero_division=0
        ),
        "Recall": recall_score(
            actual_violation, predicted_violation,
            zero_division=0
        ),
        "F1": f1_score(
            actual_violation, predicted_violation,
            zero_division=0
        ),
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn
    }

classification_rows = []

for split_name, actual, predicted in [
    ("Training", y_train, pred_train),
    ("Validation", y_val, pred_val),
    ("Test", y_test, pred_test)
]:
    for threshold in [0.0, SAFETY_MARGIN]:
        result = classification_metrics(
            actual, predicted, threshold
        )
        result["Dataset"] = split_name
        classification_rows.append(result)

classification_results = pd.DataFrame(classification_rows)

print("\n" + "=" * 70)
print("VIOLATION CLASSIFICATION")
print("=" * 70)

display(classification_results.round(4))

for threshold in [0.0, SAFETY_MARGIN]:
    print(f"\nTEST CLASSIFICATION | THRESHOLD = {threshold} ns")
    print(classification_report(
        y_test < 0,
        pred_test < threshold,
        target_names=["Timing met", "Violation"],
        zero_division=0
    ))

# ================================================================
# 7. SAVE INDIVIDUAL PREDICTIONS
# ================================================================

def make_prediction_table(df, actual, predicted, threshold):
    cols = [
        c for c in [
            "sample_id", "design", "architecture",
            "rtl_sha256", CLOCK
        ]
        if c in df.columns
    ]

    out = df[cols].copy().reset_index(drop=True)

    out["actual_wns"] = actual
    out["predicted_wns"] = predicted
    out["signed_error_ns"] = predicted - actual
    out["absolute_error_ns"] = np.abs(predicted - actual)
    out["actual_violation"] = actual < 0
    out["predicted_violation"] = predicted < threshold

    a = out["actual_violation"]
    p = out["predicted_violation"]

    out["classification"] = np.select(
        [a & p, ~a & ~p, ~a & p, a & ~p],
        ["TP", "TN", "FP", "FN"],
        default="Unknown"
    )

    return out

train_predictions = make_prediction_table(
    train, y_train, pred_train, SAFETY_MARGIN
)
val_predictions = make_prediction_table(
    val, y_val, pred_val, SAFETY_MARGIN
)
test_predictions = make_prediction_table(
    test, y_test, pred_test, SAFETY_MARGIN
)

print("\nWORST 15 TEST PREDICTIONS")
display(
    test_predictions.sort_values(
        "absolute_error_ns", ascending=False
    ).head(15).round(4)
)

print("\nMISSED TEST VIOLATIONS")
display(
    test_predictions[
        test_predictions["classification"] == "FN"
    ].round(4)
)

# ================================================================
# 8. CIRCUIT-WEIGHTED MAE
# ================================================================

def circuit_metrics(prediction_table):
    grouped = prediction_table.groupby("rtl_sha256").agg(
        architecture=("architecture", "first"),
        observations=("absolute_error_ns", "size"),
        circuit_mae=("absolute_error_ns", "mean"),
        max_error=("absolute_error_ns", "max"),
        mean_bias=("signed_error_ns", "mean")
    ).reset_index()

    return grouped

circuit_train = circuit_metrics(train_predictions)
circuit_val = circuit_metrics(val_predictions)
circuit_test = circuit_metrics(test_predictions)

print("\nCIRCUIT-LEVEL ACCURACY")

for name, table in [
    ("Training", circuit_train),
    ("Validation", circuit_val),
    ("Test", circuit_test)
]:
    print(
        f"{name:12s} | "
        f"Circuits: {len(table):3d} | "
        f"Circuit-weighted MAE: "
        f"{table['circuit_mae'].mean():.4f} ns"
    )

# ================================================================
# 9. CLOCK-REGIME ANALYSIS
# ================================================================

def clock_regime_metrics(df, actual, predicted):
    temp = pd.DataFrame({
        "clock": df[CLOCK].to_numpy(),
        "actual": actual,
        "predicted": predicted
    })

    temp["regime"] = np.select(
        [
            temp["clock"] < 3,
            (temp["clock"] >= 3) & (temp["clock"] < 7),
            np.isclose(temp["clock"], 10),
            temp["clock"] > 10
        ],
        ["T < 3 ns", "3 <= T < 7 ns", "T = 10 ns", "T > 10 ns"],
        default="Other"
    )

    rows = []

    for regime, group in temp.groupby("regime"):
        a = group["actual"].to_numpy()
        p = group["predicted"].to_numpy()

        rows.append({
            "Regime": regime,
            "Samples": len(group),
            "MAE": mean_absolute_error(a, p),
            "Mean Bias": np.mean(p - a),
            "R2": r2_score(a, p) if len(a) >= 2 else np.nan
        })

    return pd.DataFrame(rows)

print("\nTEST ACCURACY BY CLOCK REGIME")
clock_test_results = clock_regime_metrics(
    test, y_test, pred_test
)
display(clock_test_results.round(4))

# ================================================================
# 10. PLOT: ACTUAL VS PREDICTED
# ================================================================

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, name, actual, predicted in zip(
    axes,
    ["Training", "Validation", "Test"],
    [y_train, y_val, y_test],
    [pred_train, pred_val, pred_test]
):
    error = np.abs(predicted - actual)

    ax.scatter(
        actual, predicted,
        c=error, cmap="viridis",
        alpha=0.7, s=25
    )

    low = min(actual.min(), predicted.min()) - 0.5
    high = max(actual.max(), predicted.max()) + 0.5

    ax.plot([low, high], [low, high], "r--")
    ax.set_xlim(low, high)
    ax.set_ylim(low, high)

    ax.set_title(name)
    ax.set_xlabel("Actual WNS (ns)")
    ax.set_ylabel("Predicted WNS (ns)")
    ax.grid(alpha=0.3)

plt.suptitle("Random Forest: Actual vs Predicted WNS")
plt.tight_layout()
plt.savefig(
    f"{OUTPUT_DIR}/actual_vs_predicted.png",
    dpi=180, bbox_inches="tight"
)
plt.show()

# ================================================================
# 11. PLOT: RESIDUAL DISTRIBUTIONS
# ================================================================

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

for name, actual, predicted in [
    ("Training", y_train, pred_train),
    ("Validation", y_val, pred_val),
    ("Test", y_test, pred_test)
]:
    residual = predicted - actual

    axes[0].hist(
        residual, bins=35,
        alpha=0.45, label=name
    )

    axes[1].scatter(
        actual, residual,
        alpha=0.45, s=20, label=name
    )

axes[0].axvline(0, color="black", linestyle="--")
axes[0].set_title("Residual Distribution")
axes[0].set_xlabel("Prediction Error (ns)")
axes[0].set_ylabel("Count")
axes[0].legend()

axes[1].axhline(0, color="black", linestyle="--")
axes[1].set_title("Residual vs Actual WNS")
axes[1].set_xlabel("Actual WNS (ns)")
axes[1].set_ylabel("Prediction Error (ns)")
axes[1].legend()

for ax in axes:
    ax.grid(alpha=0.3)

plt.tight_layout()
plt.savefig(
    f"{OUTPUT_DIR}/residual_diagnostics.png",
    dpi=180, bbox_inches="tight"
)
plt.show()

# ================================================================
# 12. PLOT: ERROR TOLERANCE
# ================================================================

tolerances = np.linspace(0, 1.0, 101)

plt.figure(figsize=(9, 5))

for name, actual, predicted in [
    ("Training", y_train, pred_train),
    ("Validation", y_val, pred_val),
    ("Test", y_test, pred_test)
]:
    abs_error = np.abs(predicted - actual)

    accuracy = [
        100 * np.mean(abs_error <= tolerance)
        for tolerance in tolerances
    ]

    plt.plot(tolerances, accuracy, label=name)

plt.xlabel("Absolute Error Tolerance (ns)")
plt.ylabel("Predictions Within Tolerance (%)")
plt.title("WNS Prediction Accuracy vs Error Tolerance")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(
    f"{OUTPUT_DIR}/error_tolerance.png",
    dpi=180, bbox_inches="tight"
)
plt.show()

# ================================================================
# 13. PLOT: RANDOM FOREST LEARNING CURVE
# ================================================================

# Diagnostic only: fixed model parameters, no tuning.
# Grouped subsampling keeps entire RTL-hash groups together.
fractions = [0.15, 0.25, 0.4, 0.6, 0.8, 1.0]
learning_rows = []

groups = train["rtl_sha256"].to_numpy()

for fraction in fractions:

    if fraction == 1.0:
        idx = np.arange(len(train))
    else:
        splitter = GroupShuffleSplit(
            n_splits=1,
            train_size=fraction,
            random_state=SEED
        )
        idx, _ = next(
            splitter.split(X_train, D_train, groups)
        )

    # Same hyperparameters as the final model
    temp_model = Pipeline([
        ("imputer", SimpleImputer(
            strategy="median",
            add_indicator=True,
            keep_empty_features=True
        )),
        ("model", RandomForestRegressor(
            n_estimators=300,
            max_depth=12,
            min_samples_leaf=3,
            random_state=SEED,
            n_jobs=-1
        ))
    ])

    temp_model.fit(
        X_train.iloc[idx],
        D_train[idx]
    )

    train_subset_pred = predict_wns(
        temp_model,
        X_train.iloc[idx],
        T_train[idx]
    )

    validation_pred = predict_wns(
        temp_model,
        X_val,
        T_val
    )

    learning_rows.append({
        "Training samples": len(idx),
        "Training MAE": mean_absolute_error(
            y_train[idx], train_subset_pred
        ),
        "Validation MAE": mean_absolute_error(
            y_val, validation_pred
        )
    })

learning_df = pd.DataFrame(learning_rows)

plt.figure(figsize=(9, 5))

plt.plot(
    learning_df["Training samples"],
    learning_df["Training MAE"],
    "o-", label="Training MAE"
)

plt.plot(
    learning_df["Training samples"],
    learning_df["Validation MAE"],
    "o-", label="Validation MAE"
)

plt.xlabel("Number of Training Observations")
plt.ylabel("MAE (ns)")
plt.title("Random Forest Learning Curve")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(
    f"{OUTPUT_DIR}/learning_curve.png",
    dpi=180, bbox_inches="tight"
)
plt.show()

display(learning_df.round(4))

# ================================================================
# 14. PLOT: VIOLATION CONFUSION MATRICES
# ================================================================

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

for ax, threshold in zip(axes, [0.0, SAFETY_MARGIN]):

    cm = confusion_matrix(
        y_test < 0,
        pred_test < threshold,
        labels=[False, True]
    )

    ax.imshow(cm, cmap="Blues")

    for i in range(2):
        for j in range(2):
            ax.text(
                j, i, str(cm[i, j]),
                ha="center", va="center",
                fontsize=14
            )

    ax.set_xticks([0, 1], ["Met", "Violation"])
    ax.set_yticks([0, 1], ["Met", "Violation"])
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")
    ax.set_title(f"Threshold = {threshold} ns")

plt.suptitle("Test Violation Detection")
plt.tight_layout()
plt.savefig(
    f"{OUTPUT_DIR}/confusion_matrices.png",
    dpi=180, bbox_inches="tight"
)
plt.show()

# ================================================================
# 15. PLOT: TEST WNS VS CLOCK PERIOD
# ================================================================

plt.figure(figsize=(10, 6))

plt.scatter(
    T_test, y_test,
    s=30, alpha=0.7,
    label="Actual WNS"
)

plt.scatter(
    T_test, pred_test,
    marker="x", s=30,
    alpha=0.7,
    label="Predicted WNS"
)

plt.axhline(0, color="gray", linestyle="--")
plt.xlabel("Clock Period (ns)")
plt.ylabel("WNS (ns)")
plt.title("Test: WNS vs Clock Period")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(
    f"{OUTPUT_DIR}/wns_vs_clock.png",
    dpi=180, bbox_inches="tight"
)
plt.show()

# ================================================================
# 16. SAVE MODEL, METRICS AND PREDICTIONS
# ================================================================

model_bundle = {
    "model": rf_model,
    "feature_columns": structural_features,
    "target_definition": "effective_delay = clock_period - WNS",
    "clock_column": CLOCK,
    "safety_margin_ns": SAFETY_MARGIN,
    "random_state": SEED
}

joblib.dump(
    model_bundle,
    f"{OUTPUT_DIR}/physics_informed_random_forest.joblib"
)

regression_results.to_csv(
    f"{OUTPUT_DIR}/regression_metrics.csv"
)

classification_results.to_csv(
    f"{OUTPUT_DIR}/classification_metrics.csv",
    index=False
)

train_predictions.to_csv(
    f"{OUTPUT_DIR}/train_predictions.csv",
    index=False
)

val_predictions.to_csv(
    f"{OUTPUT_DIR}/validation_predictions.csv",
    index=False
)

test_predictions.to_csv(
    f"{OUTPUT_DIR}/test_predictions.csv",
    index=False
)

circuit_test.to_csv(
    f"{OUTPUT_DIR}/test_circuit_metrics.csv",
    index=False
)

clock_test_results.to_csv(
    f"{OUTPUT_DIR}/test_clock_regime_metrics.csv",
    index=False
)

learning_df.to_csv(
    f"{OUTPUT_DIR}/learning_curve.csv",
    index=False
)

print("\n" + "=" * 70)
print("FINAL MODEL SUMMARY")
print("=" * 70)

print("Algorithm: Physics-informed Random Forest")
print("Trees: 300")
print("Max depth: 12")
print("Minimum leaf samples: 3")
print("Structural features:", len(structural_features))

print("\nFINAL TEST METRICS")
print(
    pd.Series(
        regression_metrics(y_test, pred_test)
    ).round(4)
)

print("\nAll model artifacts saved to:", OUTPUT_DIR)

# ================================================================
# 17. ZIP ALL RESULTS FOR DOWNLOAD
# ================================================================

import shutil

zip_path = shutil.make_archive(
    "/content/rf_final_results",
    "zip",
    OUTPUT_DIR
)

print("Results archive:", zip_path)

from google.colab import files
files.download(zip_path)
```
